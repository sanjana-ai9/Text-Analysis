```python
text = """Python is a popular programming language.
Python is easy to learn and useful for data analysis.
Data analysis helps us understand information."""
```

```python
print(text)
```

```python
words = text.split()
```

```python
word_count = len(words)
```

```python
print("total number of words:",word_count)
```

```python
character_count = len(text)

print("Total number of characters:", character_count)
```

```python
vowels = "aeiouAEIOU"
vowel_count = 0

for char in text:
  if char in vowels:
    vowel_count += 1

print("Total number of vowels:", vowel_count)
```

```python
from collections import Counter

words = text.lower().split()
word_frequency = Counter(words)

repeated_words = {}

for word, count in word_frequency.items():
 if count > 1:
   repeated_words[word] = count

print("Repeated words:", repeated_words)
```

```python
print("TEXT ANALYSIS REPORT")
print("--------------------")
print("Total words:", len(text.split()))
print("Total characters:", len(text))

print("Total vowels:", vowel_count)
print("Repeated words:", repeated_words)
```
